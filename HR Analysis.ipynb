```python
import numpy as np
import pandas as pd
```

```python
#Load HR Dataset
hrdata = pd.read_csv('/content/HR-Employee-Attrition.csv')
hrdata
```

```python
# Check Data Types, null values
hrdata.info()
```

```python
# how many missing value in our data
hrdata.isnull().sum()
```

```python
# find duplicates values
hrdata.duplicated().sum()
```

```python
# How many employees are currently working in the company?
hrdata.shape[0]
```

```python
# What is the gender distribution of employees?
hrdata["Gender"].value_counts()
```

```python
# What is the age distribution of employees?
hrdata["Age"].describe()
```

```python
# What is the overall employee attrition rate?
hrdata["Attrition"].value_counts(normalize=True)*100
```

```python
# Which department has the highest attrition rate?
pd.crosstab(
    hrdata["Department"],
    hrdata["Attrition"],
    normalize="index"
)*100
```

```python
# Does salary influence employee attrition?
hrdata.groupby("Attrition")["MonthlyIncome"].mean()
```

```python
# Which departments pay the highest average salaries?
hrdata.groupby("Department")["MonthlyIncome"].mean()
```

```python
# Are employees leaving because of delayed promotions?
hrdata.groupby("Attrition")["YearsSinceLastPromotion"].mean()
```

```python
# How long do employees stay before leaving?
hrdata.groupby("Attrition")["YearsAtCompany"].mean()
```

```python
# Does job satisfaction affect attrition?
pd.crosstab(
    hrdata["JobSatisfaction"],
    hrdata["Attrition"],
    normalize="index"
)*100
```

```python
# Does work-life balance impact employee retention?
pd.crosstab(
    hrdata["WorkLifeBalance"],
    hrdata["Attrition"],
    normalize="index"
)*100
```

```python
# Which employee group is most at risk of leaving?
pd.crosstab(
    [hrdata["Department"], hrdata["OverTime"]],
    hrdata["Attrition"],
    normalize="index"
)*100
```
